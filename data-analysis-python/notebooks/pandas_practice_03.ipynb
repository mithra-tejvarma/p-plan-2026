#### Sorting, Ranking & Basic Statistical Operations

In this we learn about sorting , ranking and basic statistical operations using pandas

```python
import numpy as np
import pandas as pd
```

why sorting?
sorting helps us to reorder the data to reveal the patterns that are not visible in raw order

```python
df_store = pd.read_csv('../datasets/superstore_excel.csv')
df_store
```

we can sort sales row and then we can analyze ut to get the patterns

to sort the data set using values inside them we can use `df.sort_values(by,asc,inplace)`

```python
df_store.sort_values(by='Sales', ascending=False)
```

```python
df_store.sort_values(by='Profit',ascending=False)
```

```python
df_store.sort_values(by=['Region','Profit'],ascending=False)
```

To check the random data we can use `df.sample(num)`

```python
df_store.sample(7)
```

```python
df_store.sort_index(axis=1,ascending=False)
```

Ranking can just assign position numbers, It can't move the data

```python
df_store.rank()
```

rank provides the rank for the values it appear in the data set see the below example 

to use rank method syntax - `df.rank()`

```python
obj = pd.Series([7, -5, 7, 4, 2, 0, 4])
obj.rank()
```

we can give some methods so we can get the integer ranks

```python
obj.rank(method='first')#which appears first gets first rank
```

```python
obj.rank(method='first',ascending=False)
```

```python
df_store.rank().sort_values(by='Sales')
```

```python
df_store.rank().sort_values(by='Profit')
```
